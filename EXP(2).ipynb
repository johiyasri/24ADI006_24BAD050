```python
print("24BAD050-JOHIYA SRI S")
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
x = np.linspace(-10, 10, 500)

sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)

plt.figure(figsize=(12, 7))
plt.plot(x, sigmoid, label="Sigmoid")
plt.plot(x, tanh, label="Tanh")
plt.plot(x, relu, label="ReLU")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid()
plt.show()
print("\nACTIVATION FUNCTION COMPARISON")
print("Sigmoid : Output Range = (0, 1), Saturates at both ends, Can cause vanishing gradient")
print("Tanh    : Output Range = (-1, 1), Saturates at both ends, Zero-centered")
print("ReLU    : Output Range = [0, infinity), No saturation for positive values, Computationally efficient")
```

```python
print("24BAD050-JOHIYA SRI S")
import tensorflow as tf
import matplotlib.pyplot as plt
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train.reshape(-1, 784).astype("float32") / 255.0
x_test = x_test.reshape(-1, 784).astype("float32") / 255.0

activations = ["sigmoid", "tanh", "relu"]
histories = {}

for activation in activations:
    print("\nTraining ANN using", activation.upper())

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(784,)),
        tf.keras.layers.Dense(128, activation=activation),
        tf.keras.layers.Dense(64, activation=activation),
        tf.keras.layers.Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    histories[activation] = model.fit(
        x_train,
        y_train,
        validation_split=0.2,
        epochs=10,
        batch_size=128,
        verbose=1
    )

print("\nPERFORMANCE COMPARISON")
print("=" * 80)

for activation in activations:
    history = histories[activation].history
    print("\n", activation.upper())
    print("Training Accuracy   :", round(history["accuracy"][-1], 4))
    print("Validation Accuracy :", round(history["val_accuracy"][-1], 4))
    print("Training Loss       :", round(history["loss"][-1], 4))
    print("Validation Loss     :", round(history["val_loss"][-1], 4))
    print("Epochs              :", len(history["loss"]))

plt.figure(figsize=(10, 6))
for activation in activations:
    plt.plot(histories[activation].history["accuracy"], label=activation + " Train")
    plt.plot(histories[activation].history["val_accuracy"], linestyle="--", label=activation + " Validation")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(10, 6))
for activation in activations:
    plt.plot(histories[activation].history["loss"], label=activation + " Train")
    plt.plot(histories[activation].history["val_loss"], linestyle="--", label=activation + " Validation")

plt.title("Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()

best_activation = max(
    activations,
    key=lambda x: histories[x].history["val_accuracy"][-1]
)

print("\nBest Activation Function:", best_activation.upper())
```

```python
print("24BAD050-JOHIYA SRI S")
import tensorflow as tf
import matplotlib.pyplot as plt
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train = x_train.reshape(-1, 784).astype("float32") / 255.0
x_test = x_test.reshape(-1, 784).astype("float32") / 255.0
optimizers = {
    "SGD": tf.keras.optimizers.SGD(learning_rate=0.01),
    "Momentum": tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
    "RMSProp": tf.keras.optimizers.RMSprop(learning_rate=0.001),
    "Adam": tf.keras.optimizers.Adam(learning_rate=0.001)
}

histories = {}

for name, optimizer in optimizers.items():
    print("\nTraining ANN using", name)

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(784,)),
        tf.keras.layers.Dense(128, activation="relu"),
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    histories[name] = model.fit(
        x_train,
        y_train,
        validation_split=0.2,
        epochs=10,
        batch_size=128,
        verbose=1
    )

print("\nOPTIMIZER PERFORMANCE COMPARISON")
print("=" * 80)

for name in optimizers:
    history = histories[name].history

    print("\n", name)
    print("Training Loss       :", round(history["loss"][-1], 4))
    print("Validation Loss     :", round(history["val_loss"][-1], 4))
    print("Training Accuracy   :", round(history["accuracy"][-1], 4))
    print("Validation Accuracy :", round(history["val_accuracy"][-1], 4))

plt.figure(figsize=(10, 6))

for name in optimizers:
    plt.plot(histories[name].history["loss"], label=name)

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(10, 6))

for name in optimizers:
    plt.plot(histories[name].history["val_loss"], label=name)

plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(10, 6))

for name in optimizers:
    plt.plot(histories[name].history["accuracy"], label=name + " Train")
    plt.plot(histories[name].history["val_accuracy"], linestyle="--", label=name + " Validation")

plt.title("Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()

best_optimizer = max(
    optimizers.keys(),
    key=lambda x: histories[x].history["val_accuracy"][-1]
)

print("\nBest Optimizer Based on Validation Accuracy:", best_optimizer)
```
