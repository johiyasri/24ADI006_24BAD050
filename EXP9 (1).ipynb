```python
print("24BAD050-JOHIYA SRI S")
!pip install -q transformers torch sentencepiece
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

prompt = "What is Artificial Intelligence?"
inputs = tokenizer(prompt, return_tensors="pt")
outputs = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_new_tokens=100,
    temperature=0.7,
    do_sample=True
)
response = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("User Prompt:", prompt)
print("Generated Response:", response)
```

```python
print("24BAD050-JOHIYA SRI S")
def chatbot_response(query, history=None):
    conversation = ""
    if history:
        for message in history[-4:]:
            if message["role"] == "user":
                conversation += "User: " + message["content"] + "\n"
            elif message["role"] == "assistant":
                conversation += "Assistant: " + message["content"] + "\n"

    prompt = conversation + "User: " + query + "\nAssistant:"
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    outputs = model.generate(
        inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=100,
        temperature=0.7,
        do_sample=True
    )

    response = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return response
queries = [
    "What is Machine Learning?",
    "Explain Deep Learning.",
    "What is Python?"
]
for query in queries:
    print("User:", query)
    print("Bot:", chatbot_response(query))
    print()
```

```python
print("24BAD050-JOHIYA SRI S")
!pip install -q gradio
import gradio as gr
def chat_interface(message, history):
    response = chatbot_response(message, history)
    return response
demo = gr.ChatInterface(
    fn=chat_interface,
    title="AI Chatbot",
    description="Interactive Chatbot using a Pre-trained Transformer LLM",
    examples=[
        "What is Artificial Intelligence?",
        "Explain Machine Learning.",
        "What is Deep Learning?",
        "What is Python?"
    ]
)
demo.launch(share=True)
```

```python
print("24BAD050-JOHIYA SRI S")

questions = [
    "What is Artificial Intelligence? Explain in simple words.",
    "Explain neural networks in simple words.",
    "What is Natural Language Processing? Explain with an example."
]

for question in questions:
    response = chatbot_response(question)
    print("\nQuestion:", question)
    print("Response:", response)

print("\nChatbot testing completed!")
print("Requirements file created successfully!")
print("Experiment completed successfully!")
```
